## **التدفق العام للمشروع**


### 1. **الإعدادات** → `config.py`

* يحتوي كل الإعدادات (بيانات، نموذج، تدريب، مسارات، جهاز)
* أي ملف يحتاج معلومة (زي عدد epochs أو مكان حفظ النموذج) يستوردها من هنا

---

#### 2. **المساعدة العامة** → `src/utils/data_utils.py`

* يوفر دوال مساعده:

  * `get_device()` → يحدد إذا نستخدم GPU أو CPU.
  * `seed_everything()` → يثبت العشوائية ( reproducibility).
  * `build_transforms()` → تجهيز الصور (augmentations).
  * `create_dataloaders()` → يبني dataloaders للتدريب والاختبار.
  * `print_dataset_info()` → يطبع إحصائيات عن البيانات.

---

### 3. **تجهيز البيانات** → `src/data/prepare_data.py`

* يستورد الدوال من `data_utils.py`.
* يتأكد أن مجلدات `train/` و `val/` موجودة.
* يبني dataloaders ويطبع تفاصيل البيانات.
* يرجّع:

  * `dataloaders`
  * `dataset_sizes`
  * `class_names`

---

### 4. **النموذج** → `src/models/model.py`

* يعرّف كلاس `ImageClassifierModel`.
* يبني ResNet18 مع تعديل الطبقة الأخيرة.
* يحتوي على دالة `train_model()` اللي تنفذ التدريب (loop كامل).
* مسؤول عن:

  * تحميل البيانات (من utils).
  * تجهيز النموذج.
  * التدريب وحفظ أفضل وزن.

---

### 5. **التدريب الأساسي** → `main.py`

* **نقطة البداية** للمشروع:

  1. يحمّل الإعدادات من `config.py`.
  2. يضبط reproducibility (`seed_everything`).
  3. يجهز البيانات (`prepare_data`).
  4. يبني النموذج (`create_model`).
  5. يحدد الجهاز (`get_device`).
  6. يدرب النموذج (`train_model`).
  7. يحفظ النموذج (`save_model`).
* في النهاية: يولّد ملف `ants_bees_model.pth`.

---

### 6. **الاختبار** → `test.py`

* يستخدم النموذج المدرب للتنبؤ على صور جديدة.
* خطواته:

  1. تحميل الأوزان من `ants_bees_model.pth`.
  2. تجهيز الصورة (transforms).
  3. تمريرها للنموذج.
  4. طباعة النتيجة (التصنيف + الثقة).
  5. عرض الصورة مع التنبؤ.

---

## 🔄 Flow Chart (تدفق البيانات)⏬

```


                config.py (الإعدادات)
                         |        
                         ▼        
                src/utils/data_utils.py (دوال مساعدة عامة)
                         │
                         ▼
                src/data/prepare_data.py (تجهيز البيانات)
                         │
                         ▼
                src/models/model.py (بناء وتدريب النموذج)
                         │
                         ▼
                main.py (تنفيذ التدريب + حفظ النموذج)
                         │
                         ▼
                ants_bees_model.pth (ملف الأوزان الناتج)
                         │
                         ▼
                test.py (تحميل النموذج → تجربة على صور جديدة)
```

---

## ✅ النتيجة النهائية

* **main.py** = تدريب النموذج.
* **test.py** = اختبار النموذج.
* **config.py** = كل الإعدادات في مكان واحد.
* **utils** = أدوات عامة (جهاز، random seed، dataloaders...).
* **data** = تجهيز البيانات.
* **models** = تعريف وتدريب النموذج.
